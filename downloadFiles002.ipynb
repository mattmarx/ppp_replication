```python
import requests
import os

def download_zenodo_files(record_id, file_names, output_dir=".", access_token=None):
    """
    Download specified files from a Zenodo repository.
    
    Parameters:
    - record_id: Zenodo record ID (e.g., '15058362')
    - file_names: List of file names to download
    - output_dir: Directory to save files (default: current directory)
    - access_token: API token for restricted files (optional)
    """
    # Construct API URL
    base_url = "https://zenodo.org/api/records"
    url = f"{base_url}/{record_id}"
    
    # Set headers with token if provided
    headers = {"Authorization": f"Bearer {access_token}"} if access_token else {}
    
    # Fetch record metadata
    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        raise Exception(f"Failed to fetch metadata: {response.status_code} - {response.text}")
    
    record = response.json()
    files = record.get('files', [])
    if not files:
        raise Exception("No files found in the record.")
    
    # Fail loudly if a requested file is not in this record
    missing = set(file_names) - {f['key'] for f in files}
    if missing:
        raise Exception(f"Not in Zenodo record {record_id}: {sorted(missing)}")
    
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    
    # Download specified files
    for file_info in files:
        file_name = file_info['key']
        if file_name in file_names:
            file_url = file_info['links']['self']
            file_path = os.path.join(output_dir, file_name)
            
            expected_size = file_info.get("size")
            if os.path.exists(file_path) and expected_size is not None and os.path.getsize(file_path) == expected_size:
                print(f"Skipping (exists, size matches): {file_name}")
                continue
            print(f"Downloading: {file_name}")
            file_response = requests.get(file_url, headers=headers, stream=True)
            if file_response.status_code == 200:
                with open(file_path, 'wb') as f:
                    for chunk in file_response.iter_content(chunk_size=8192):
                        f.write(chunk)
                print(f"Saved: {file_path}")
            else:
                print(f"Failed to download {file_name}: {file_response.status_code}")

```
